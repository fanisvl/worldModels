# Latent Traversal

```python
import torch
import numpy as np
import matplotlib.pyplot as plt


def traverse_latent(model_path, dim_max_idx, latent_dim, num_steps=15, span=(-3, 3)):
    model = VAE(latent_dim=latent_dim, inverted_colors=True).to('cpu')
    model.load_state_dict(torch.load(model_path, map_location='cpu'))
    model.eval()

    z_base = torch.zeros((1, latent_dim), device='cpu')
    values = np.linspace(span[0], span[1], num_steps)
    
    for dim_idx in range(dim_max_idx):
        fig, axes = plt.subplots(1, num_steps, figsize=(num_steps * 2.5, 4))
        for i, val in enumerate(values):
            z = z_base.clone()
            z[0, dim_idx] = val
            with torch.no_grad():
                img = model.decode(z).cpu().squeeze(0).permute(1, 2, 0).numpy()
            axes[i].imshow(img)
            axes[i].axis("off")
        plt.show()
```

```python
import torch
from torch.utils.data import DataLoader
import numpy as np
from data.dataset import RolloutDataset

def compute_latent_quantile_spans(model, dataset, latent_dim, q_low=0.05, q_high=0.95, batch_size=2, device='cpu'):
    # compute per dimension quantiles of the latent means over the dataset
    # returns arrays of shape [latent_dim] for low and high values
    model.eval()
    loader = DataLoader(dataset, batch_size=batch_size, shuffle=False)
    
    all_mu = []
    with torch.no_grad():
        for batch in loader:
            obs = batch[0].to(device)
            mu, log_var = torch.split(model.encoder(obs), latent_dim, dim=1)
            all_mu.append(mu.cpu())
    
    all_mu = torch.cat(all_mu, dim=0).numpy() # [num_samples, latent_dim]
    lows = np.quantile(all_mu, q_low, axis=0)
    highs = np.quantile(all_mu, q_high, axis=0)
    return lows, highs


model = VAE(latent_dim=64, inverted_colors=True).to('cpu')
model.load_state_dict(torch.load('../models/vae/vae.lat64.e100.bs128.inv.vizdoom_1_32_2.5k.14-08_epoch25.pt', map_location='cpu'))
train_dataset = RolloutDataset(data_dir='/media/fanis/Intenso/wm18/vizdoom_val', max_samples=10)

lows, highs = compute_latent_quantile_spans(model, train_dataset, latent_dim=64, device='cpu')

# Now for dimension j you can use:

dim_j = 0
span_j = (lows[dim_j], highs[dim_j])
print(f"Latent dim {dim_j} span: {span_j}")
```

```python
import matplotlib.pyplot as plt

def plot_latent_histograms(model, dataset, latent_dim, batch_size=64, device='cuda'):
    model.eval()
    loader = DataLoader(dataset, batch_size=batch_size, shuffle=False)
    all_mu = []

    with torch.no_grad():
        for batch in loader:
            obs = batch[0].to(device)
            mu, log_var = torch.split(model.encoder(obs), latent_dim, dim=1)
            all_mu.append(mu.cpu())
    
    all_mu = torch.cat(all_mu, dim=0).numpy()  # shape [num_samples, latent_dim]

    # Plot histograms
    fig, axes = plt.subplots(latent_dim // 4 + 1, 4, figsize=(16, 3 * (latent_dim // 4 + 1)))
    axes = axes.flatten()
    for i in range(latent_dim):
        axes[i].hist(all_mu[:, i], bins=30, color='skyblue', edgecolor='black')
        axes[i].set_title(f'Latent dim {i}')
    plt.tight_layout()
    plt.show()

plot_latent_histograms(model, train_dataset, 64, device='cpu')
```

```python
import sys
sys.path.append('..')
from data.dataset import LatentSequenceDataset
from modules.vae import VAE
import torch
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
import ipywidgets as widgets
from IPython.display import display, clear_output
!jupyter nbextension enable --py widgetsnbextension

# --- 1. SETUP: Load VAE model and Latent Dataset ---
device = 'cpu'
latent_dim = 64
n_pca_components = 32
VAE_MODEL_PATH = '../models/vae/vae.lat64.e100.bs128.inv.vizdoom_1_32_2.5k.14-08_epoch25.pt'
LATENT_DATA_DIR = '/media/fanis/Intenso/wm18/latent64_vizdoom_1_32_2.5k'

print("Loading VAE model (we only need the decoder)...")
model = VAE(latent_dim=latent_dim, inverted_colors=True).to(device)
model.load_state_dict(torch.load(VAE_MODEL_PATH, map_location=device))
model.eval()

# print("\nLoading pre-computed latent dataset...")
# latent_dataset = LatentSequenceDataset(data_dir=LATENT_DATA_DIR)

# --- 2. EXTRACT ALL LATENT VECTORS ---
print("\nExtracting all latent vectors from the dataset...")
all_latents_list = [ep['latents'] for ep in latent_dataset.episodes]
all_latents_tensor = torch.cat(all_latents_list, dim=0)
all_latents_np = all_latents_tensor.numpy()

print(f"Extracted a total of {all_latents_np.shape[0]} latent vectors.")

# # --- 3. FIT PCA and Analyze ---
print(f"\nFitting PCA on all latent vectors with {n_pca_components} components...")
pca = PCA(n_components=latent_dim)
pca.fit(all_latents_np)

# Check explained variance
explained_variance = pca.explained_variance_ratio_
print(f"Top {n_pca_components} Principal Components:")
for i in range(min(n_pca_components, len(explained_variance))):
    print(f"PC{i+1}: {explained_variance[i]:.2%}")
print(f"Total variance explained by top {n_pca_components} PCs: {np.sum(explained_variance[:n_pca_components]):.2%}")

# --- 4. CREATE INTERACTIVE VISUALIZATION WITH IMPROVED UI ---
print("\nSetting up interactive visualization...")

# Transform data into the PCA space to find sensible slider ranges
latent_pca_space = pca.transform(all_latents_np)
pc_lows = np.quantile(latent_pca_space, 0.05, axis=0)
pc_highs = np.quantile(latent_pca_space, 0.95, axis=0)

# The mean vector will be our starting point
mean_pca_vector = np.mean(latent_pca_space, axis=0)

# --- UI WIDGET CREATION ---

# Create sliders
sliders = []
for i in range(n_pca_components):
    slider = widgets.FloatSlider(
        value=mean_pca_vector[i],
        min=pc_lows[i],
        max=pc_highs[i],
        step=0.05, # A finer step can be nice
        description=f'PC {i+1}',
        style={'description_width': 'initial'},
        layout=widgets.Layout(width='95%'), # Use percentage for responsiveness
        continuous_update=False # Only update on release for performance
    )
    sliders.append(slider)

# Create output widget for the plot
output_plot = widgets.Output()

# Create a reset button
reset_button = widgets.Button(
    description='Reset to Mean',
    button_style='info',
    tooltip='Reset all sliders to their mean value'
)

# --- UI EVENT HANDLERS ---

def update_plot(*args):
    """Decodes the current slider state and displays the image."""
    with output_plot:
        clear_output(wait=True)

        # Create a PC vector from current slider values
        current_pc_vector = mean_pca_vector.copy()
        for i, slider in enumerate(sliders):
            current_pc_vector[i] = slider.value

        # Transform back to original latent space
        reconstructed_latent_z = pca.inverse_transform(current_pc_vector.reshape(1, -1))

        # Decode with VAE
        with torch.no_grad():
            latent_tensor = torch.from_numpy(reconstructed_latent_z).float().to(device)
            generated_img_tensor = model.decode(latent_tensor)

            img = generated_img_tensor.squeeze().cpu().numpy()
            img = np.transpose(img, (1, 2, 0))

            plt.figure(figsize=(6, 6))
            plt.imshow(np.clip(img, 0, 1))
            plt.title("VAE Decoder Output")
            plt.axis('off')
            plt.show()

def on_reset_button_clicked(b):
    """Event handler to reset sliders to their mean values."""
    # Temporarily un-observe to prevent triggering updates for every single slider change
    for slider in sliders:
        slider.unobserve(update_plot, names='value')
    
    # Set all sliders back to the mean
    for i, slider in enumerate(sliders):
        slider.value = mean_pca_vector[i]
        
    # Re-observe and trigger a single update
    for slider in sliders:
        slider.observe(update_plot, names='value')
    update_plot() # Manually trigger one update after resetting


# Attach observers to sliders and button
for slider in sliders:
    slider.observe(update_plot, names='value')
reset_button.on_click(on_reset_button_clicked)


# --- UI LAYOUT ASSEMBLY ---

# Split sliders into two columns for a more compact layout
num_sliders = len(sliders)
mid_point = (num_sliders + 1) // 2
col1 = widgets.VBox(sliders[:mid_point])
col2 = widgets.VBox(sliders[mid_point:])

# Create a control panel with the button and slider columns
# Add some margin to the right of the panel for spacing
control_panel = widgets.VBox([
    reset_button,
    widgets.HBox([col1, col2])
], layout=widgets.Layout(margin='0px 30px 0px 0px'))

# Create the final UI by placing the control panel and the output side-by-side
title = widgets.HTML("<h3>Interactive VAE Latent Space Explorer</h3>")
ui = widgets.VBox([
    title,
    widgets.HBox([control_panel, output_plot])
])

# Display the interface
display(ui)

# Trigger the initial plot
update_plot()
```

# UMAP Visualization

```python
import umap
import torch
import numpy as np
import matplotlib.pyplot as plt
from bokeh.plotting import figure, show, output_notebook
from bokeh.models import HoverTool, ColumnDataSource, CustomJS, Div
from bokeh.layouts import row, column
import base64
from io import BytesIO
from PIL import Image
output_notebook()  # configure bokeh for notebook output

MAX_POINTS_UMAP = 5000
N_INTERACTIVE_POINTS = 2500 # to be decoded for visualization
RANDOM_SEED = 32

# Reduce UMAP size for memory 
if all_latents_np.shape[0] > MAX_POINTS_UMAP:
    print(f"Subsampling {MAX_POINTS_UMAP} points for UMAP fitting...")
    fit_indices = np.random.RandomState(RANDOM_SEED).choice(all_latents_np.shape[0], MAX_POINTS_UMAP, replace=False)
    fit_latents = all_latents_np[fit_indices]
else:
    fit_latents = all_latents_np

print("Fitting UMAP on sampled latent vectors...")
reducer = umap.UMAP(n_neighbors=10, min_dist=0.1, n_components=2, random_state=RANDOM_SEED)
embedding = reducer.fit_transform(fit_latents)
print(f"UMAP fitting complete. Embedding shape: {embedding.shape}")

# plot with latent decoding
def create_interactive_plot_with_decoding():
    """Creates a plot that displays pre-decoded images (via JS) when hovered."""
    # Subsample for the interactive plot
    if len(embedding) > N_INTERACTIVE_POINTS:
        interactive_indices = np.random.RandomState(RANDOM_SEED).choice(len(embedding), N_INTERACTIVE_POINTS, replace=False)
        plot_embedding = embedding[interactive_indices]
        plot_latents = torch.tensor(fit_latents[interactive_indices])
    else:
        plot_embedding = embedding
        plot_latents = torch.tensor(fit_latents)
    
    source = ColumnDataSource(data=dict(
        x=plot_embedding[:, 0],
        y=plot_embedding[:, 1],
        index=list(range(len(plot_embedding)))
    ))
    
    p = figure(
        width=600,
        height=600,
        tools="pan,wheel_zoom,box_zoom,reset,save",  # Remove 'tap' tool
        title="UMAP Projection - Hover over points to see decoded images"
    )
    
    # Make points more visible - no selection styling needed for hover
    p.scatter('x', 'y', source=source, size=8, alpha=0.7, color='blue')
    
    return p, source, plot_latents

interactive_plot, interactive_source, interactive_latents = create_interactive_plot_with_decoding()


# --- Pre-decode images in Python and attach as base64 strings to the data source ---
print("Decoding interactive images in batches...")
latents = interactive_latents.to(device).float()

imgs_b64 = []
batch_size = 64
with torch.no_grad():
    for i in range(0, latents.shape[0], batch_size):
        decoded = model.decode(latents[i:i+batch_size]).cpu()  # [B, C, H, W]
        for img_t in decoded:
            img_np = img_t.permute(1, 2, 0).numpy()
            img_np = np.clip(img_np, 0, 1)
            arr = (img_np * 255).astype(np.uint8)
            # Handle grayscale or RGB
            if arr.ndim == 3 and arr.shape[2] == 1:
                pil = Image.fromarray(arr[..., 0], mode='L')
            else:
                pil = Image.fromarray(arr)
            buf = BytesIO()
            pil.save(buf, format='PNG')
            imgs_b64.append(base64.b64encode(buf.getvalue()).decode('ascii'))

if len(imgs_b64) != len(interactive_source.data['x']):
    print(f"Warning: number of images ({len(imgs_b64)}) != number of points ({len(interactive_source.data['x'])})")

# Attach images to CDS
interactive_source.data = dict(
    x=interactive_source.data['x'],
    y=interactive_source.data['y'],
    index=interactive_source.data['index'],
    img_b64=imgs_b64,
)

# --- JS callback to render hovered image in a Div ---
img_div = Div(width=600, height=620, text="""
  <div><b>Hover over a point to see its decoded image.</b></div>
""")

# Create custom HoverTool with JavaScript callback
hover_cb = CustomJS(args=dict(src=interactive_source, div=img_div), code="""
  // Get the hovered index from the hover tool
  const indices = cb_data.index.indices;
  if (!indices || indices.length === 0) { 
    return; 
  }
  
  const i = indices[0];
  const x = src.data.x[i].toFixed(2);
  const y = src.data.y[i].toFixed(2);
  const b64 = src.data.img_b64[i];
  
  div.text = `
    <div><b>Index:</b> ${i} &nbsp; (<b>X</b> ${x}, <b>Y</b> ${y})</div>
    <img src="data:image/png;base64,${b64}" style="width:580px;height:auto;border:1px solid #ccc;margin-top:6px;" />
  `;
""")

hover_tool = HoverTool(
    tooltips=None,
    callback=hover_cb
)
interactive_plot.add_tools(hover_tool)

print("Hover over points in the plot below to see decoded images:")
show(row(interactive_plot, img_div))
print(f"\nInteractive plot ready! Hover over any point to show its decoded image.")
print(f"Available points: 0 to {len(interactive_source.data['index'])-1}")
```
